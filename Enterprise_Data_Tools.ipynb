```python
import pandas as pd
from pathlib import Path
```

```python
DATA_DIR = Path("data/csvs")

print(DATA_DIR)
```

```python
employees_df = pd.read_csv(
    DATA_DIR / "employees_with_null.csv"
)

leave_df = pd.read_csv(
    DATA_DIR / "leave_balance.csv"
)

expenses_df = pd.read_csv(
    DATA_DIR / "expense_records.csv"
)

it_assets_df = pd.read_csv(
    DATA_DIR / "IT_assets.csv"
)

offices_df = pd.read_csv(
    DATA_DIR / "office_locations.csv"
)

print("All datasets loaded successfully!")
```

```python
print("Employees:", employees_df.shape)
print("Leave:", leave_df.shape)
print("Expenses:", expenses_df.shape)
print("IT Assets:", it_assets_df.shape)
print("Offices:", offices_df.shape)
```

```python
import pandas as pd
import glob
import os

# Get all CSV files
csv_files = glob.glob("data/csvs/*.csv")

for file in csv_files:
    
    # Read CSV
    df = pd.read_csv(file)

    print("\n" + "=" * 70)
    print(f"FILE: {os.path.basename(file)}")
    print("=" * 70)

    # Basic information
    print(f"\nShape: {df.shape}")
    print(f"Rows: {df.shape[0]}")
    print(f"Columns: {df.shape[1]}")

    print("\nColumn Names:")
    print(list(df.columns))

    # Data types
    print("\nData Types:")
    print(df.dtypes)

    # Missing values
    print("\nMissing Values:")
    print(df.isnull().sum())

    # Duplicate rows
    print("\nDuplicate Rows:")
    print(df.duplicated().sum())

    # Preview
    print("\nFirst 3 Rows:")
    print(df.head(3))

print("\n" + "=" * 70)
print("DATA VALIDATION COMPLETED")
print("=" * 70)
```
