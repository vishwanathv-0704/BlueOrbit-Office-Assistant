```python
def get_employee_details(employee_id):
    """
    Retrieve employee information using employee ID.
    """

    result = employees_df[
        employees_df["employee_id"] == employee_id
    ]

    if result.empty:
        return {
            "success": False,
            "message": f"Employee {employee_id} not found."
        }

    employee = result.iloc[0]

    return {
        "success": True,
        "employee_id": employee["employee_id"],
        "name": employee["name"],
        "department": employee["department"],
        "designation": employee["designation"],
        "manager_id": employee["manager_id"],
        "location_id": employee["location_id"],
        "joining_date": employee["joining_date"],
        "employment_status": employee["employment_status"]
    }
```

```python
def get_leave_balance(employee_id):
    """
    Retrieve remaining leave balance for an employee.
    """

    result = leave_df[
        leave_df["employee_id"] == employee_id
    ]

    if result.empty:
        return {
            "success": False,
            "message": f"Leave information for {employee_id} not found."
        }

    leave = result.iloc[0]

    return {
        "success": True,
        "employee_id": employee_id,
        "casual_leave": int(leave["casual_leave"]),
        "earned_leave": int(leave["earned_leave"]),
        "sick_leave": int(leave["sick_leave"])
    }
```

```python
get_leave_balance("EMP001")
```

```python
def get_expenses(employee_id):
    """
    Retrieve expense records and expense summary
    for an employee.
    """

    result = expenses_df[
        expenses_df["employee_id"] == employee_id
    ].copy()

    if result.empty:
        return {
            "success": True,
            "employee_id": employee_id,
            "expense_count": 0,
            "total_amount": 0,
            "approved_amount": 0,
            "pending_amount": 0,
            "rejected_amount": 0,
            "expenses": []
        }

    result["amount"] = pd.to_numeric(
        result["amount"],
        errors="coerce"
    ).fillna(0)

    total_amount = result["amount"].sum()

    approved_amount = result.loc[
        result["status"] == "Approved",
        "amount"
    ].sum()

    pending_amount = result.loc[
        result["status"] == "Pending",
        "amount"
    ].sum()

    rejected_amount = result.loc[
        result["status"] == "Rejected",
        "amount"
    ].sum()

    return {
        "success": True,
        "employee_id": employee_id,
        "expense_count": len(result),
        "total_amount": float(total_amount),
        "approved_amount": float(approved_amount),
        "pending_amount": float(pending_amount),
        "rejected_amount": float(rejected_amount),
        "expenses": result.to_dict(orient="records")
    }
```

```python
get_expenses("EMP010")
```

```python
def get_it_assets(employee_id):
    """
    Retrieve IT assets assigned to an employee.
    """

    result = it_assets_df[
        it_assets_df["employee_id"] == employee_id
    ]

    if result.empty:
        return {
            "success": True,
            "employee_id": employee_id,
            "asset_count": 0,
            "assets": []
        }

    return {
        "success": True,
        "employee_id": employee_id,
        "asset_count": len(result),
        "assets": result.to_dict(orient="records")
    }
```

```python
get_it_assets("EMP001")
```

```python
def get_office_location(location_id):
    """
    Retrieve office information using location ID.
    """

    result = offices_df[
        offices_df["location_id"] == location_id
    ]

    if result.empty:
        return {
            "success": False,
            "message": f"Location {location_id} not found."
        }

    office = result.iloc[0]

    return {
        "success": True,
        "location_id": office["location_id"],
        "city": office["city"],
        "office_name": office["office_name"],
        "address": office["address"],
        "facilities": office["facilities"],
        "working_hours": office["working_hours"]
    }
```

```python
def validate_data():
    """
    Perform basic validation on all enterprise datasets.
    """

    validation = {
        "employee_rows": len(employees_df),
        "leave_rows": len(leave_df),
        "expense_rows": len(expenses_df),
        "asset_rows": len(it_assets_df),
        "office_rows": len(offices_df),

        "employee_missing_values":
            int(employees_df.isnull().sum().sum()),

        "leave_missing_values":
            int(leave_df.isnull().sum().sum()),

        "expense_missing_values":
            int(expenses_df.isnull().sum().sum()),

        "asset_missing_values":
            int(it_assets_df.isnull().sum().sum()),

        "office_missing_values":
            int(offices_df.isnull().sum().sum()),

        "duplicate_employee_ids":
            int(employees_df["employee_id"].duplicated().sum()),

        "duplicate_expense_ids":
            int(expenses_df["expense_id"].duplicated().sum()),

        "duplicate_asset_ids":
            int(it_assets_df["asset_id"].duplicated().sum()),

        "duplicate_location_ids":
            int(offices_df["location_id"].duplicated().sum())
    }

    return validation
```
